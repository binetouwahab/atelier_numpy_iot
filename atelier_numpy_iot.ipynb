Partie 1 : Listes Python vs tableaux NumPy
On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8
1) Créer une liste Python stockant ces températures;

```python
liste = [24.5, 25.1, 26.3, 27.0, 25.8]
liste
```

2) Créer un tableau NumPy équivalent stockant ces températures ;

```python
import numpy as np

tab = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
tab
```

3) Effectuer une multiplication par 2 sur ces deux collections;

```python
mult_liste = liste * 2
mult_liste
```

```python
mult_tab = tab * 2
mult_tab
```

Quelle différence observe-t-on ? 

on constate que avec les liste ca a dipluquer la liste alors que avec le tableau ca a appliquer la multiplication
